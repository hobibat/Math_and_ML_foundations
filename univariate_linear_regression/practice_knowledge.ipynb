```python
w= 100
b=100
```

```python
def compute_model_output(x, w, b):
    f_wb=np.zeros(len(x))
    #made a zero array of lenght x_train array

    for i in range(m):
        f_wb[i]= w*x[i] + b
    return f_wb
```

```python
ret_fwb= compute_model_output(x_train, w, b)
```

```python
plt.plot(x_train, ret_fwb, c='b', label='our prediction')
# so this function takes two arguments
#the first is the arrays of the x points 
#the other is an array of the y points we have
#and the function connects the points together
#in our case, we want to connect the x features with the y hat

plt.scatter(x_train, y_train, marker='x', c='r', label='acutal values') 
#and this line to show the actual points on the same graph for the predicted
#so we can see the difference
```
